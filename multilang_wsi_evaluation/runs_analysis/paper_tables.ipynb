```python
import typing as tp
from pathlib import Path

import numpy as np
import pandas as pd
```

```python
metrics_dir = Path("../../outputs/2022-08-13_11-21-25/GLMVectorizer/agg_metrics")
```

```python
def prepare_df(file_name: str, rounding: tp.Optional[int] = 1) -> pd.DataFrame:
    file_path = metrics_dir / file_name
    metric_df = pd.read_csv(file_path, sep='\t')
    if rounding is not None:
        metric_df['score'] = metric_df['score'].apply(lambda score: round(score, rounding))
    return metric_df
```
